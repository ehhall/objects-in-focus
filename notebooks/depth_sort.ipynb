```python
# imports
import numpy as np
import cv2
import os
import matplotlib.pyplot as plt
from PIL import Image, ImageDraw
from scipy.ndimage import binary_dilation

# functions
def load_numpy_arrays(file_paths):
    """
    Takes a list of file paths to numpy arrays and returns a list of the loaded arrays.

    Parameters:
    file_paths (list of str): A list of file paths to numpy arrays.

    Returns:
    list of numpy.ndarray: A list containing the loaded numpy arrays.
    """

    loaded_arrays = [np.load(file) for file in file_paths]
    return loaded_arrays

def load_images(image_paths):
    images = []
    for path in image_paths:
        try:
            img = Image.open(path)
            images.append(img)
        except Exception as e:
            print(f"Could not load image at {path}: {e}")
    return images

def list_files(path):
    """Lists all files in the specified directory."""

    files = []
    for entry in os.listdir(path):
        if os.path.isfile(os.path.join(path, entry)):
            files.append(entry[:-4])
    return files

def make_mask_test_images(image_list, mask_list):
    for i, image in enumerate(image_list):

        # Set up test image
        test_image = np.zeros_like(image)

        # Get unique mask IDs
        mask_array = mask_list[i]
        unique_masks = np.unique(mask_array)

        for mask_id in unique_masks:
          # Create boolean mask for the current region
          mask = (mask_array == mask_id)

          # Expand mask to 3 channels
          mask_3d = np.expand_dims(mask, axis=2)
          mask_3d = np.repeat(mask_3d, 3, axis=2)
          masked_image = np.where(mask_3d == 1, image, 0)

          # Add masked image to test_image
          test_image = test_image + masked_image

          # Create an outline of the mask
          dilated_mask = binary_dilation(mask, iterations=1)  # Dilation
          outline = dilated_mask & ~mask  # Subtract original mask from dilated mask

          # Add the outline to the test image (e.g., red outline)
          outline_color = [255, 0, 0]  # Red color
          for c in range(3):  # Apply color to each channel
              test_image[:, :, c] = np.where(outline, outline_color[c], test_image[:, :, c])

        # Save the test image to a folder for review
        final_image = Image.fromarray(test_image.astype(np.uint8))
        image_id = str(i)
        print(f"Processing image {i}")
        final_image.save(f"/content/drive/MyDrive/projects/CVPR/test_mask_images/{image_id}.png")

# set dir
img_dir = 'images/'
mask_dir = 'masks/'

# get image paths
image_list = list_files(img_dir)
image_paths = [img_dir + item + '.png' for item in image_list]
mask_paths = [mask_dir + item + '.npy' for item in image_list]

# load masks and images to lists
mask_list = load_numpy_arrays(mask_paths)
image_list = load_images(image_paths)

# Call the function with image and mask lists
make_mask_test_images(image_list, mask_list)
```
